3. Write a Python function that takes a list of IPL team names and prints only those teams whose names are longer than 6 characters, skipping the rest using the continue statement.


```python
iplteam = ["royal", "chennai super" , "gujarat titan" , "RCB", "pbks", "laucknow"]

for i in iplteam:
    if len(i)<6:
        continue
    print(i)
```

5. Build a simple shopping cart total calculator: Given a list of item prices from a Flipkart cart, use a loop to sum the prices. If an item price is 0 (out of stock), skip it. Stop adding items if the running total crosses ₹2000
using break, and print the final total .< br><br><em><strong>Constraint :< /strong> Use both break and continue in your solution .< /em>



```python
price = [100,200,3000,400,000,1200,3000,21212,12,324,4234,344,5235,354345,43545,435345,45345,345,5434,45]
sum = 0
for i in price:
    if i == 0:
        continue
    elif (sum + i)>20000:
        break
    else:
        sum = sum + i
print(sum)    
```
